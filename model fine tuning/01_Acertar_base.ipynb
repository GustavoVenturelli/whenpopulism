```python
import pandas as pd
```

```python
base_train = pd.read_csv('train_total_stratified.csv')
base_test = pd.read_csv('test_total_stratified.csv')
nova_base = pd.concat([base_train,base_test])
nova_base.drop(columns="Unnamed: 0", axis=1,inplace=True)
print(len(nova_base))
nova_base.drop_duplicates(subset='text', keep='first', inplace=True)
nova_base.drop_duplicates(subset='index_banco', keep='first', inplace=True)
print(len(nova_base))

nova_base.head()
```

```python
nova_base.groupby(['ae', 'pc', 'pop']).size().reset_index(name='contagem')

```

```python
sample_teste = nova_base.sample(520,random_state=57)
sample_teste.groupby(['ae', 'pc', 'pop']).size().reset_index(name='contagem')
```

```python
sample_treino = nova_base[~nova_base["index_banco"].isin(sample_teste["index_banco"])]
sample_treino.groupby(['ae', 'pc', 'pop']).size().reset_index(name='contagem')
```

```python
print(len(sample_teste) + len(sample_treino))
print(sum(sample_treino['text'].isin(sample_teste['text'])))

#sample_teste.to_csv('test_total_stratified v2.csv',index=False)
#sample_treino.to_csv('train_total_stratified v2.csv',index=False)
```

```python
filtro = (sample_treino['pop'] == 0) & (sample_treino['ae'] == 0) & (sample_treino['pc'] == 0)
selecionados_class0  = sample_treino[filtro].sample(720, random_state=22)
filtro2 = (sample_treino['pop'] == 1) | (sample_treino['ae'] == 1) | (sample_treino['pc'] == 1)


selecionados_class1 = sample_treino[filtro2]
```

```python
subamostragem = pd.concat([selecionados_class0,selecionados_class1])
#subamostragem.to_csv('train_sub_stratified v2.csv',index=False)
```
