```python
import os
import pandas as pd
import numpy as np
```

```python
result_files = [x for x in os.listdir("model") if x.endswith("csv")]
result_files
```

```python
base = pd.read_csv(os.path.join("model",result_files[0]))
base = pd.DataFrame(
                   columns=base.columns)

for i in result_files:
    print(i)
    base = pd.concat([base, pd.read_csv(os.path.join("model",i))], ignore_index=True)

```

```python
base.to_excel("Consolidado v3.xlsx",index=False)
```
